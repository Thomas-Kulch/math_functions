### Rolling Mean Function

```python
def rolling_mean(arr,win_size):

    '''
    This function will take in a list of numbers, calculate the moving average of that list, 
    adjust the window size, and plot the output on a line graph in matplotlib
    '''

    i = 0
    #initialize empty list to store rolling means
    rolling_mean = []

    #loop through array to consider every window
    while i < len(arr) - win_size + 1:
        
        #store elements from i to i+win_size in a list to get the current window
        window = arr[i:i + win_size]

        #calculate the average of current window
        window_average = round(sum(window) / win_size,2)

        #store the average of current window in rolling mean list
        rolling_mean.append(window_average)

        #increment window
        i += 1

    #plot the array
    plt.plot(rolling_mean)
    plt.xlabel('Index')
    plt.ylabel('Rolling Mean')
    plt.title('Rolling Mean Plot')
    plt.show()

    return plt
```

```python
#create array to test function
arr = [1,2,3,6,7,9,12,13,14,20,21,25,26,27,28,29,31,2,5,10]

rolling_mean(arr,3)
```

### Basic Stats Function

```python

def statistics(arr):
    '''
    This function will take a list of numbers, calculate the mean, median, mode, variance, standard deviation and print it out.
    '''

    #sort the list for median calculation
    sorted_arr = sorted(arr)
    n = len(sorted_arr)

    #calculate median
    if n % 2 == 1:
        #odd
        med_indx = n // 2
        med = sorted_arr[med_indx]
    else:
        #even
        med_indx_1 = n // 2 - 1
        med_indx_2 = n // 2
        med = (sorted_arr[med_indx_1] + sorted_arr[med_indx_2]) / 2

    #calculate mean
    mean = sum(arr) / len(arr)
    
    #calculate mode
    mode = None
    max_count = 0
    for i in range(len(arr)):
        count = arr.count(i)
        if count > max_count:
            mode = i
            max_count = count

    #calculate variance. we have mean already
    #calculate the sum of squared differences from the mean
    sum_sq_diff = sum((x - mean) ** 2 for x in arr)

    var = sum_sq_diff / len(arr)

    #calculate standard deviation
    std_dev = np.sqrt(var)


    print("Mean:", mean)
    print("Median:", med)
    print("Mode:", mode)
    print("Variance:", var)
    print("Standard Deviation:", std_dev)
```

```python
#test function using same array from earlier
statistics(arr)
```

### Vector Multiplication Function

```python

def dot_product(arr1,arr2):
    '''
    This function will calculate the dot product of two vectors without using numpy function.

    Vectors can only be multiplied if they are the same length. 
    Use try/except to print an error if the inputted arrays are not the same length.
    '''
    
    try:
        result = 0
        for i in range(len(arr1)):
            result += arr1[i] * arr2[i]
        return print("Dot Product:",result)
    except:
        print(f'Error: Vectors must be the same size. Vector 1 has a length of {len(arr1)} and Vector 2 has a length of {len(arr2)}')
```

```python
# testing vectors of different lengths
arr1 = [1,2,3,4,5,6]
arr2 = [1,2,3,4]
dot_product(arr1,arr2)
```

```python
# testing vectors of same lengths
arr1 = [1,3,5,7,9]
arr2 = [2,4,6,8,10]
dot_product(arr1,arr2)
```

### Matrix Multiplication Function

```python
def matrix_multi(matrix1,matrix2):
    '''
    This function will take two matrices, check if they are able to be multiplied with each other, 
    check if they are invertible, multiplies them if possible and returns an error if not.
    '''
    try:
        result = []
        
        for i in range(len(matrix1)):
            row = []
            for j in range(len(matrix2[0])):
                sum = 0
                for k in range(len(matrix2)):
                    sum += matrix1[i][k] * matrix2[k][j]
                row.append(sum)
            result.append(row)

        return print("Matrix Product:",result)
    except:
        print(f'Error: The number of columns in the first matrix must equal the number of rows in the second matrix.')
```

```python
# test with matrices
matrix1 = [[1, 2, 3],
           [4, 5, 6]]

matrix2 = [[7, 8],
           [9, 10],
           [11, 12]]

matrix_multi(matrix1,matrix2)
```
